```python
# Importing libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [120000, 145000, 138000, 165000, 190000, 215000]
```

```python
months = [1, 2, 3, 4, 5, 6]
campaign_a = [50, 65, 72, 90, 110, 125]
campaign_b = [45, 60, 68, 75, 95, 105]

plt.plot(months, campaign_a, marker="o", color="blue", label="Campaign A")
plt.plot(months, campaign_b, marker="s", color="orange", linestyle="--", label="Campaign B")

plt.xlabel("Month")
plt.ylabel("Customers Acquired")
plt.title("Customer Acquisition Comparison: Campaign A vs Campaign B")

plt.xlim(1, 6)

plt.legend()
plt.grid(True, linestyle=":", alpha=0.6)

plt.show()
```

```python
months = [1, 2, 3, 4, 5, 6]
patients = [420, 450, 480, 510, 495, 540]
waiting_time = [65, 62, 58, 55, 60, 49]
satisfaction = [72, 74, 76, 78, 77, 83]

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 12))

ax1.plot(months, patients, marker="o", color="blue")
ax1.set_title("Monthly Patient Volume")
ax1.set_xlabel("Month")
ax1.set_ylabel("Number of Patients")
ax1.grid(True, linestyle=":", alpha=0.6)

ax2.plot(months, waiting_time, marker="o", color="orange")
ax2.set_title("Average Waiting Time")
ax2.set_xlabel("Month")
ax2.set_ylabel("Waiting Time (Minutes)")
ax2.grid(True, linestyle=":", alpha=0.6)

ax3.plot(months, satisfaction, marker = "o" , color="green")
ax3.set_title("Patient Satisfaction Score")
ax3.set_xlabel("Month")
ax3.set_ylabel("Satisfaction (%)")
ax3.grid(True, linestyle=":", alpha=0.6)

fig.tight_layout()
plt.show()
```

```python
months = [1, 2, 3, 4, 5, 6]
revenue = [150000, 165000, 158000, 190000, 215000, 240000]
customers = [80, 92, 88, 110, 125, 140]
product_a = [45, 50, 48, 60, 68, 75]
product_b = [35, 40, 42, 45, 52, 58]

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 12))

ax1.plot(months, revenue, marker="o", color="tab:blue", linewidth=2)
ax1.set_title("Monthly Revenue Trend")
ax1.set_xlabel("Month")
ax1.set_ylabel("Revenue (₦)")
ax1.grid(True, linestyle=":", alpha=0.6)

ax2.plot(months, customers, marker="s", color="tab:green", linewidth=2)
ax2.set_title("Customer Growth")
ax2.set_xlabel("Month")
ax2.set_ylabel("Active Customers")
ax2.grid(True, linestyle=":", alpha=0.6)

ax3.plot(months, product_a, marker="o", color="tab:purple", label="Product A")
ax3.plot(months, product_b, marker="^", color="tab:orange", linestyle="--", label="Product B")
ax3.set_title("Product Performance Comparison")
ax3.set_xlabel("Month")
ax3.set_ylabel("Units Sold")
ax3.legend()
ax3.grid(True, linestyle=":", alpha=0.6)

fig.tight_layout()
plt.show()
```
